## **Teoría del Confort Térmico en Ganado Vacuno**

El confort térmico en ganado vacuno se refiere a un estado en el que los animales no experimentan estrés por calor ni por frío, permitiéndoles mantener su temperatura corporal central dentro de un rango óptimo con el mínimo gasto energético. Este estado es crucial para la salud, el bienestar y la productividad del ganado, especialmente en la producción lechera y de carne.

Factores clave que influyen en el confort térmico incluyen:

*   **Temperatura Ambiental:** La temperatura del aire circundante.
*   **Humedad Relativa:** La cantidad de vapor de agua en el aire.
*   **Radiación Solar:** La exposición directa o indirecta a la luz solar.
*   **Velocidad del Viento:** El movimiento del aire, que puede ayudar a la disipación del calor.
*   **Precipitación:** Lluvia o nieve, que pueden afectar la temperatura corporal.

El **Índice de Temperatura y Humedad (ITH)** es una herramienta ampliamente utilizada para evaluar el nivel de estrés por calor en el ganado. Combina la temperatura de bulbo seco y la temperatura de bulbo húmedo para proporcionar una medida única del confort térmico. Valores más altos de ITH indican un mayor riesgo de estrés por calor.

El estrés por calor en el ganado vacuno puede manifestarse a través de diversos síntomas, como:

*   Aumento de la frecuencia respiratoria (jadeo).
*   Reducción del consumo de alimento.
*   Disminución de la producción de leche o del aumento de peso.
*   Incremento de la temperatura corporal.
*   Cambios en el comportamiento (búsqueda de sombra, aumento del consumo de agua).

Mantener a los animales dentro de su zona de confort térmico es esencial para optimizar su rendimiento productivo y reproductivo, así como para garantizar su bienestar general.

# **Metodos y calculos**
* Indice de temperatura y humedad (produccion intensiva)


$TH = 0.72(T{bs} +T{bs})+40.6 $

donde:
1.   $ T _ {bs}$ Temeperatura bulbo seco ($°C$)
2.   $T_{hb}$ Temperatura del bulto humedo ($°C$)


Para la correlación de los datos de la especie bovina en la producción de leche se recomienda utilizar los límites propuestos por National Weather Service:

1. Si $ITH <= 70$ ambiente no estresante
2. S1 $ITH > 70 \space and space ITH <= 78$
situación de alerta por estrés térmico
3. Si ITH 78 0 ≤ 83 peligro por estrés térmico
4. S1 ITH 83 situación critica




```python
# ENTRADA: pedir tbs (Temperatura de Bulbo seco) y tbh (Temperatura de bulbo hümedo)
tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))

# PROCESO: Calculo de ith (Indice de Temperatura y Humedad
ith = 0.72 * (tbs + tbh) + 40.6

# SALIDA: Imprimir resultado, de la evaluación del ith y el confort térmico
if ith <= 70:
  print("El ambiente es no estresante")
elif ith > 70 and ith <= 78:
  print("La situación es de alerta por estrés térmico")
elif ith > 78 and ith  <= 83:
  print("La situación es de  peligro por estrés térmico")
else:
  print("La situación es critica")
```

## Explicación del Código de Confort Térmico

Este código Python calcula el Índice de Temperatura y Humedad (ITH) y clasifica el nivel de estrés térmico basándose en los valores proporcionados. A continuación, se detalla cada parte:

1.  **`tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))`**
    *   Pide al usuario que introduzca la temperatura de bulbo seco en grados Centígrados.
    *   La función `input()` lee la entrada como una cadena de texto.
    *   `float()` convierte esa cadena de texto a un número decimal (flotante) y lo almacena en la variable `tbs`.

2.  **`tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))`**
    *   De manera similar, pide al usuario la temperatura de bulbo húmedo en grados Centígrados.
    *   Convierte la entrada a un número flotante y la guarda en la variable `tbh`.

3.  **`ith = 0.72 * (tbs + tbh) + 40.6`**
    *   Esta línea calcula el Índice de Temperatura y Humedad (ITH) utilizando la fórmula proporcionada:
        `ITH = 0.72 * (Tbs + Tbh) + 40.6`.
    *   El resultado se almacena en la variable `ith`.

4.  **`if ith <= 70:`**
    *   Esta es la primera condición de una estructura `if-elif-else`.
    *   Verifica si el valor de `ith` es menor o igual a 70.
    *   Si es verdadero, se imprime el mensaje indicando que el ambiente `Es no estresante`.

5.  **`elif ith > 70 and ith <= 78:`**
    *   Si la condición anterior fue falsa, esta `elif` (else if) verifica si `ith` es mayor que 70 Y menor o igual que 78.
    *   Si ambas condiciones son verdaderas, se imprime `La situación es de alerta por estrés térmico`.

6.  **`elif ith > 78 and ith <= 83:`**
    *   Si las condiciones anteriores fueron falsas, esta `elif` verifica si `ith` es mayor que 78 Y menor o igual que 83.
    *   Si ambas son verdaderas, se imprime `La situación es de peligro por estrés térmico`.

7.  **`else:`**
    *   Si ninguna de las condiciones `if` o `elif` anteriores fue verdadera, significa que `ith` es mayor que 83.
    *   En este caso, se ejecuta el bloque `else`, imprimiendo `La situación es critica`.
